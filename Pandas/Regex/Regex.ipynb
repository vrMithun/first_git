# Regex full guide

```python
import pandas as pd
```

### Basic matching

```python
df = pd.DataFrame({
    'notes': [
        'apple pie is tasty',
        'banana bread is sweet',
        'apple grapes juice is healthy',
        'grapes are fresh',
        'pineapple tart',
        'abcdefg'
    ]
})

#going to find the data rows containing the word apple

display(df[df["notes"].str.contains("apple",case=False,na=False)])
```

### Extracting substrings

```python
df["fruits"]=df["notes"].str.extract(r'(apple|banana|grapes|pineapple)')
display(df.head())
```

* The above code only finds the first valid string and retun it

```python
df["fruits"]=df['notes'].str.findall(r'(apple|banana|grapes|pineapple)')
display(df)
```
