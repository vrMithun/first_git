```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
```

```python
data=[
  {"Hours_Study": 1, "Hours_Sleep": 7, "Passed": 0},
  {"Hours_Study": 2, "Hours_Sleep": 6, "Passed": 0},
  {"Hours_Study": 3, "Hours_Sleep": 8, "Passed": 0},
  {"Hours_Study": 4, "Hours_Sleep": 5, "Passed": 1},
  {"Hours_Study": 5, "Hours_Sleep": 6, "Passed": 1},
  {"Hours_Study": 6, "Hours_Sleep": 5, "Passed": 1},
  {"Hours_Study": 7, "Hours_Sleep": 7, "Passed": 1},
  {"Hours_Study": 8, "Hours_Sleep": 4, "Passed": 1}
]

df=pd.DataFrame(data)
display(df.head())
```

```python
x=df[["Hours_Study","Hours_Sleep"]]
target=df["Passed"]
xtrain,xtest,ytrain,ytest=train_test_split(x,target,test_size=0.3,random_state=42)
```

```python
model=SVC()
model.fit(xtrain,ytrain)
```

```python
prediction=model.score(xtest,ytest)
print(prediction)
```

```python
ypred=model.predict(xtest)
```

```python
model.predict([[1,8]])
```

```python
from sklearn.metrics import confusion_matrix
```

```python
cm=confusion_matrix(ytest,ypred)
```

```python
import seaborn as sns
```

```python
sns.heatmap(cm,annot=True)
```
