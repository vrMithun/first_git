```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import Perceptron
```

```python
data=[
  {"Hours_Study": 1, "Hours_Sleep": 8, "Passed": 0},
  {"Hours_Study": 2, "Hours_Sleep": 7, "Passed": 0},
  {"Hours_Study": 3, "Hours_Sleep": 6, "Passed": 0},
  {"Hours_Study": 4, "Hours_Sleep": 5, "Passed": 1},
  {"Hours_Study": 5, "Hours_Sleep": 5, "Passed": 1},
  {"Hours_Study": 6, "Hours_Sleep": 6, "Passed": 1},
  {"Hours_Study": 7, "Hours_Sleep": 4, "Passed": 1},
  {"Hours_Study": 8, "Hours_Sleep": 3, "Passed": 1},
  {"Hours_Study": 2, "Hours_Sleep": 9, "Passed": 0},
  {"Hours_Study": 9, "Hours_Sleep": 2, "Passed": 1}
]

df=pd.DataFrame(data)
display(df.head())
```

```python
from sklearn.model_selection import train_test_split
```

```python
x=df[["Hours_Study","Hours_Sleep"]]
target=df["Passed"]
xtrain,xtest,ytrain,ytest=train_test_split(x,target,test_size=0.2,random_state=42)
```

```python
model=Perceptron(eta0=0.1,random_state=42)
model.fit(xtrain,ytrain)
```

```python
model.score(xtest,ytest)
```

```python
ypred=model.predict(xtest)
```

```python
from sklearn.metrics import confusion_matrix
cm=confusion_matrix(ytest,ypred,labels=[0,1])
```

```python
import seaborn as sns
sns.heatmap(cm,annot=True)
```
