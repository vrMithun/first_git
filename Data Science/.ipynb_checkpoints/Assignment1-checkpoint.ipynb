### 3.Rename columns: replace spaces with ’_’ to allow dot notation


```python
def underscore(mystr):
    word_list=list(mystr.split(" "))
    result=""
    for i in range(len(word_list)):
        if i<(len(word_list)-1) and (i!=len(word_list)-1):
            result+=word_list[i]
            result+="_"
        else:
            result+=word_list[i]
    return result
columns=list(df.columns)
for i in columns:
    df.rename(columns={i:underscore(i.strip())},inplace=True)
display(df.head())
```

### 14.Oversample houses with over 10 rooms


```python
# Step 1: Filter houses with more than 10 rooms
oversample_df = df[df["ROOMS"] > 10]

# Step 2: Concatenate this subset multiple times to oversample
# Example: duplicate them 3 times
oversample_df = pd.concat([oversample_df]*3, ignore_index=True)

# Step 3: Concatenate the oversampled rows with the original DataFrame
df_oversampled = pd.concat([df, oversample_df], ignore_index=True)
```
